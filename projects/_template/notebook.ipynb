````python
# Setup: find the course, load the check helper, and print the lanes.
import json
import sys
import urllib.error
import urllib.request
from pathlib import Path

ROOT = Path.cwd()
while not (ROOT / "pyproject.toml").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from bootcamp_agent.bonus import BONUS, bonus

# write this: import your checks module, which registers the checks, for example
# from bootcamp_agent.projects import your_project  # noqa: F401

PROJECT = ROOT / "projects" / "_template"  # write this: your project folder
MODEL = ""  # write this: the Ollama model the live lane needs, or leave empty
RECORDED_ON = ""  # write this: the date of the run saved in data/recorded/


def model_is_pulled(name: str) -> bool:
    """True when a local Ollama server answers and has this model."""
    try:
        with urllib.request.urlopen("http://localhost:11434/api/tags", timeout=2) as response:
            return any(m["name"].startswith(name) for m in json.loads(response.read())["models"])
    except (urllib.error.URLError, OSError):
        return False


def check_step(check_id: str, value: object) -> None:
    """Run a project check and print its verdict. Until it is registered, say so."""
    if check_id in BONUS:
        bonus(check_id, value)
    else:
        print(f"{check_id}: not registered yet. See projects/_template/README.md, step 5.")


LIVE = bool(MODEL) and model_is_pulled(MODEL)
print(f"model: [live] {MODEL}" if LIVE else f"model: [recorded] {RECORDED_ON or 'no recording yet'}")
print(f"project: {PROJECT.relative_to(ROOT)}")
````
