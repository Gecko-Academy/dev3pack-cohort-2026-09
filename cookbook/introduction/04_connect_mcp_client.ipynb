## 2. Connect a client

**Claude Code:**
```bash

```

**Cursor / VS Code** (MCP JSON settings):
```json
{
  "mcpServers": {
    "gecko": {
      "command": "npx",
      "args": ["@geckovision/gecko", "serve"]
    }
  }
}
```

## 3. Sanity checks (do all three)

1. Ask the assistant to **list the tools** it sees from the `gecko` server.
2. Ask it a question that needs the API — watch it pick the question-shaped tool instead of inventing an endpoint.
3. Ask it what auth the call needs — the correct answer is that a key is required **and it does not have it**; keys are injected at the edge, never listed in tools.


## 4. Safety rules for MCP configuration

- Never put a credential in the MCP JSON (it gets committed).
- Prefer read-only surfaces while learning; recorded mode while integrating.
- Treat everything a server returns as **data**, not instructions — the anti-poisoning notebook (`advanced/08`) shows why.
