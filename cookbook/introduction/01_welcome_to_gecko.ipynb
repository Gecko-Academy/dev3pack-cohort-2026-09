# 01 — Welcome to Gecko

**Goal:** understand what Gecko is and run its doctor.

**Gecko** ([gecko-surf](https://github.com/GeckoVision/gecko-surf)) is the **API comprehension layer for agents**: point it at an API surface — an OpenAPI spec, a docs page, or a Solana program — and it turns that surface into question-shaped, first-call-correct agent tools, served over MCP, with **provenance** on what it claims and a **recorded mode** that costs $0.

The three ideas to keep:

1. **Comprehension, not guessing** — the agent gets the surface's real structure (a graph), not a vibe about endpoint names.
2. **Provenance** — every claim says where it came from: declared, recovered, measured, or inferred.
3. **Recorded before live** — everything can be exercised offline before any call counts.

**Prerequisites:** none for the offline cells; Node.js + `npx` for the live cells.

````python
# Live-mode guard: cells that invoke the real gecko CLI run only when you
# opt in AND npx is available. Everything else in this notebook is offline.
#   export GECKO_COOKBOOK_LIVE=1   # to enable live cells
import os
import shutil

GECKO_LIVE = os.environ.get("GECKO_COOKBOOK_LIVE") == "1" and shutil.which("npx") is not None
FIXTURES = None
from pathlib import Path
here = Path.cwd()
while not (here / "pyproject.toml").exists():
    here = here.parent
FIXTURES = here / "cookbook" / "fixtures"
print(f"live mode: {GECKO_LIVE} — fixtures: {FIXTURES}")
````

## The CLI at a glance

```bash
npx @geckovision/gecko doctor    # environment check
npx @geckovision/gecko add <spec-or-url>   # comprehend a surface
npx @geckovision/gecko report    # what was understood, with provenance
npx @geckovision/gecko serve     # expose it over MCP
```

The flow is deliberately shaped like this course's habits: check the environment (doctor), comprehend (add), **inspect the evidence** (report), then connect (serve).

````python
import subprocess

if GECKO_LIVE:
    result = subprocess.run(
        ["npx", "@geckovision/gecko", "doctor"],
        capture_output=True, text=True, timeout=300,
    )
    print(result.stdout[-3000:] or result.stderr[-3000:])
else:
    print("SKIP (offline): set GECKO_COOKBOOK_LIVE=1 with npx installed to run the doctor.")
````
