## 2. The broken import

**Context.** The cell below fails. It is the single commonest import mistake in a `src/` layout:
the module is inside a package, and the import does not say so.

**Instructions.** Run it, read the error, fix the import. Do not change anything else.

````python
def load_the_corpus():
    # `documents` is a module INSIDE the bootcamp_agent package, so the import
    # has to name the package. Nothing called `documents` sits on sys.path.
    from bootcamp_agent.documents import load_corpus

    return load_corpus(REPO_ROOT / "data" / "corpus")


try:
    print(f"{len(load_the_corpus())} documents")
except ModuleNotFoundError as error:
    print(f"❌ {error}  <- read this, then fix the import above")
````

**Expected output**

```
6 documents
✅ w02-e3 passed
```

````python
check("w02-e3", load_the_corpus)
````

## 3. A docstring that earns its place

**Context.** A caller has two questions a signature cannot answer: what comes back, and what
blows up. A docstring that repeats the function name answers neither.

**Instructions.** Write a docstring for `load_corpus`. Summary line first, ending in a full
stop. Then say what it returns and what it raises.

````python
docstring = """Read every Markdown file in `directory` into a list of Document.

Returns the documents in filename order, so retrieval is reproducible across
machines. Raises CorpusError when the directory is missing or holds no
Markdown, because an empty corpus makes every later answer a silent refusal.
"""
print(docstring)
````

**Expected output**

```
Read every Markdown file in `directory` into a list of Document.
...
✅ w02-e2 passed
```

````python
check("w02-e2", docstring)
````
