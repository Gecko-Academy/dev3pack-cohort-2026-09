# Start here

Welcome. This notebook is the map: run the cell below and it lists every part of the
course in the order you should do it, with a link to each one and a tick against what you
have finished.

**If you have not set up yet**, close this and follow [`SETUP.md`](SETUP.md) first. It
takes about half an hour and ends with a green doctor.

**Where to begin.** Week 0 is open now and is self-paced. Its four fast-lane units are the
floor session 1 assumes, about three hours in total. The three courses beside them go
deeper and nothing waits for you to finish them.

**How each notebook works.** Read the context, edit the line marked
`# <------ EDIT THIS LINE`, then run the `check(...)` cell. It prints ✅ or ❌ with the fix
named. `review("ch03")` at the end is your scorecard. Stuck? `hint("ch03-e1")` costs 30
marks and `hint("ch03-e1", reveal=True)` costs 70, out of 100. Nothing is locked, and the
number is there to show you where to spend another ten minutes rather than to rank anyone.

````python
# Reads the curriculum live, so anything added to the course shows up here by itself.
import sys
from pathlib import Path

REPO_ROOT = Path.cwd()
while not (REPO_ROOT / "pyproject.toml").exists() and REPO_ROOT != REPO_ROOT.parent:
    REPO_ROOT = REPO_ROOT.parent
sys.path.insert(0, str(REPO_ROOT / "src"))

from IPython.display import Markdown, display

from bootcamp_agent.curriculum import (
    CAPSTONE,
    CHAPTERS,
    WEEK0_COURSES,
    WEEK0_UNITS,
    WEEK_TITLES,
    exercise_ids,
    unit_exercise_ids,
)
from bootcamp_agent.hints import all_attempts

passed = {a.exercise for a in all_attempts() if a.passed}


def mark(ids: tuple[str, ...]) -> str:
    """How far through an item you are, from your own machine's progress file."""
    if not ids:
        return "&nbsp;&nbsp;&nbsp;"
    done = sum(1 for i in ids if i in passed)
    symbol = "✅" if done == len(ids) else ("▶" if done else "·")
    return f"{symbol} {done}/{len(ids)}"


lines = ["## Week 0 — start here, self-paced, offline", ""]
for course in WEEK0_COURSES:
    units = [u for u in WEEK0_UNITS if u.course == course]
    lines += [f"**{course}**", ""]
    for unit in units:
        ids = unit_exercise_ids(unit.prefix)
        link = unit.notebook.relative_to(REPO_ROOT)
        lines.append(f"- {mark(ids)} &nbsp; [{unit.number}. {unit.title}]({link})")
    lines.append("")

for module in sorted(WEEK_TITLES):
    lines += [f"## Week {module} — {WEEK_TITLES[module]}", ""]
    for chapter in (c for c in CHAPTERS if c.module == module):
        ids = exercise_ids(chapter.chapter_id) if chapter.has_notebook else ()
        title = f"{chapter.number}. {chapter.title}"
        when = chapter.on.strftime("%a %d %b")
        if chapter.notebook is None:
            lines.append(f"- &nbsp;&nbsp;&nbsp; &nbsp; {title} — *{when}, no notebook*")
        else:
            link = chapter.notebook.relative_to(REPO_ROOT)
            note = " *(needs an assistant open)*" if chapter.manual_reason else ""
            lines.append(f"- {mark(ids)} &nbsp; [{title}]({link}) — *{when}*{note}")
    lines.append("")

# The capstone has no date, because it is built in the hours between sessions.
capstone_ids = unit_exercise_ids(CAPSTONE.prefix)
capstone_link = CAPSTONE.notebook.relative_to(REPO_ROOT)
lines += [
    "## The capstone — built between sessions, from week 2",
    "",
    f"- {mark(capstone_ids)} &nbsp; [{CAPSTONE.title}]({capstone_link})",
    "",
    "## Then, whenever you like",
    "",
    "- [Project workspaces](workspaces/) — open-ended, ungraded, one per week plus a news digest",
    "- [Gecko cookbook](cookbook/) — ten notebooks on making an API agent-usable",
    f"- [Depth track](depth/) — {len(sorted(p for p in (REPO_ROOT / 'depth').iterdir() if p.is_dir()))}"
    " optional modules on the software engineering underneath",
    "- [Final assignment](final_assignment/) — graded on a private set, with a signed certificate",
]

total = len(passed)
display(Markdown("\n".join(lines)))
print(f"\n{total} exercises passed so far. Progress is a file on this machine only: ~/.bootcamp/progress.db")
````
